## Summary of Units 1-5

Let's quickly recap the journey:

1.  **Unit 1: Introduction & EDA:**
    *   Loaded the Bank Marketing data using Fireducks.
    *   Performed initial inspection (`.head()`, `.info()`, `.describe()`).

    *   Explored numerical correlations (and noted the 'duration' issue).

2.  **Unit 2: Data Cleaning & Feature Engineering:**
    *   Handled 'unknown' values (via mode imputation).

4.  **Unit 4: Data Splitting for Evaluation:**
